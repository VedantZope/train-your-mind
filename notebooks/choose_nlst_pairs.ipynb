```python
import pandas as pd
nlst = pd.read_csv("/data/vedant/MIND/benchmarks/nlst/greedy_registration_mind_masked/keypoint_distances.csv")
nlst.describe()
```

```python
nlst["percentage improvement"] = (nlst["dist_initial_physical"] - nlst["dist_final_physical"])/nlst["dist_initial_physical"]
```

```python
# sort nlst by percentage improvement and take top 50 pairs
nlst_sorted = nlst.sort_values(by="percentage improvement", ascending=True)
top_50 = nlst_sorted.head(50)
top_50_csv = top_50[["case", "fixed_image", "moving_image"]]
top_50_csv.to_csv("/data/vedant/MIND/notebooks/nlst_choosen_pairs/worst_50.csv", index=False)
```
